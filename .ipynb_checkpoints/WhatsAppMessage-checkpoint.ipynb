```python
import pywhatkit as kit
import time
import socket
from datetime import datetime

numero = "+22379790709"
message = """Bonjour Princesse ❤️
Comment tu vas ?
Je te souhaite une excellente journée.
Que le Seigneur nous garde.
Prends soin de toi.
Je t'aime beaucoup !"""

heure_envoi = 8
minute_envoi = 16

message_envoye_aujourdhui = False


def connexion_internet():
    try:
        socket.create_connection(("8.8.8.8", 53), timeout=5)
        return True
    except OSError:
        return False


while True:
    maintenant = datetime.now()

    # Reset du statut chaque jour à minuit
    if maintenant.hour == 0 and maintenant.minute == 0:
        message_envoye_aujourdhui = False

    # Si on est à l'heure OU légèrement après (marge de 5 minutes)
    if (
        maintenant.hour == heure_envoi
        and minute_envoi <= maintenant.minute <= minute_envoi + 5
        and not message_envoye_aujourdhui
    ):

        if connexion_internet():
            print("Internet OK. Envoi du message...")
            kit.sendwhatmsg_instantly(numero, message, wait_time=15)
            message_envoye_aujourdhui = True
            print("Message envoyé avec succès ✅")
            time.sleep(60)
        else:
            print("Pas de connexion Internet. Nouvelle tentative dans 30s...")

    time.sleep(30)
```
